Find the tallest indexed building and save a 256 px chip.
Run this notebook from `sidelook_v2/` or from `examples/`.
The lite sample has no building of 30 m or more (tallest is 29.8 m), so this takes the tallest row rather than filtering.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "index").is_dir() and (ROOT.parent / "index").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import sidelook

ds = sidelook.open(ROOT)
tall = ds.buildings().sort_values("height_m", ascending=False).iloc[0]
chip = ds.chip(tall.building_id, size=256)
ds.plot_chip(chip, out=ROOT / "examples" / "tallest.png")
print(tall.building_id, round(float(tall.height_m), 1), "m", "partial" if chip["partial"] else "full")
```
